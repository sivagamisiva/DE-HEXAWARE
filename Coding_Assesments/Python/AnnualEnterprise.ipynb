## Step 2: Import required libraries and load dataset


```python
import pandas as pd

df = pd.read_csv("assesment_dataset.csv")
```

```python
# Preview the first few rows
df.head()
```

## Step 3: Data Cleaning

### 1. Basic Info and Null Value Check

```python
# Get structure of the dataset
df.info()

# Check for missing/null values
df.isnull().sum()
```

### 2. Descriptive Statistics

```python
# Show descriptive statistics for numeric columns
df.describe()
```

### 3. Drop Duplicate Rows (if any)

```python
# Check and drop duplicate rows
print("Duplicates before:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates after:", df.duplicated().sum())
```

## Step 4:Spliting Dataset

```python
# DataFrame 1: Basic industry information (unique values)
df1 = df[['Industry_code_NZSIOC', 'Industry_name_NZSIOC', 'Units']].drop_duplicates()

# DataFrame 2: Financial variables (many-to-one relationship with industry)
df2 = df[['Industry_code_NZSIOC', 'Variable_name', 'Value', 'Year']]
```

##  Step 5: Perform Joins Between df1 and df2

```python
df1['Industry_code_NZSIOC'] = df1['Industry_code_NZSIOC'].astype(str)
df2['Industry_code_NZSIOC'] = df2['Industry_code_NZSIOC'].astype(str)
```

### 5.1 Inner Join (only matching records)

```python
inner_join = pd.merge(df2, df1, on='Industry_code_NZSIOC', how='inner')
print("Inner Join Result:")
inner_join.head(25)
```

### 5.2 Left Join (all from df2 + matched from df1)

```python
left_join = pd.merge(df2, df1, on='Industry_code_NZSIOC', how='left')
print("Left Join Result:")
left_join.head(30)
```

### 5.3 Right Join (all from df1 + matched from df2)

```python
right_join = pd.merge(df2, df1, on='Industry_code_NZSIOC', how='right')
print("Right Join Result:")
right_join.head(30)
```

### 5.4 Outer Join (all records from both)

```python
outer_join = pd.merge(df2, df1, on='Industry_code_NZSIOC', how='outer')
print("Outer Join Result:")
outer_join.head(30)
```

## Save & Export Merged Data

```python
inner_join.to_csv("merged_data.csv", index=False)
from google.colab import files
files.download("merged_data.csv")
```
