```python
from pyspark import SparkContext
from pyspark.sql import SparkSession
# Create SparkContext and SparkSession
sc = SparkContext.getOrCreate()
spark = SparkSession.builder.appName("PySpark DataFrame from RDD").getOrCreate()
rdd = sc.parallelize([('C', 85, 76, 87, 91),('A', 85, 78, 96, 92),('A', 92, 76, 89, 96) ], 4)
print(type(rdd))
```

## Example program

```python
sub = ['Division','English','Mathematics','Physics','Chemistry']
marks_df= spark.createDataFrame(rdd, schema=sub)
print(type(marks_df))
marks_df.printSchema()
marks_df.show()
```

## Reading csv file

```python
# Basic CSV read (no header, no schema inference)
df = spark.read.csv("/content/Marks_data.csv")
df.show()  # shows raw data

# CSV read with header and schema inference
df1 = spark.read.csv("/content/Marks_data.csv", header=True, inferSchema=True)
df1.show()  # shows parsed data with headers

# Print schema and data types
df1.printSchema()
print(df1.dtypes)
```

## Reading JSON File

```python
# Read JSON file
df2 = spark.read.option("multiLine", True).json("/content/Product details.json")
df2.show(25)





# Print schema
df2.printSchema()

# Print data types
print(df2.dtypes)
```

## Reading text file

```python
df = spark.read.text("/content/Product queries.txt")

# Rename the column from default 'value' to 'query'
df = df.withColumnRenamed("value", "query")

# Show the content neatly
df.show(truncate=False)

# Print schema to confirm structure
df.printSchema()

# Print column names and types
print(df.dtypes)



```

## Pyspark Dataframe to pandas Dataframe

```python
# Read each CSV separately
fish_df = spark.read.csv('/content/Fish.csv', header=True, inferSchema=True)
salary_df = spark.read.csv('/content/salary_data.csv', header=True, inferSchema=True)

# Show PySpark dataframes
print("Fish DataFrame:")
fish_df.show()


print("Salary DataFrame:")
salary_df.show()


fish_pd = fish_df.toPandas()
salary_pd = salary_df.toPandas()

print("\nPandas Fish DataFrame:")
print(fish_pd.head())

print("\nPandas Salary DataFrame:")
print(salary_pd.head())





```

```python
from pyspark.sql.functions import lit
```

```python

data = [["1", "sravan", "company 1"],
        ["2", "ojaswi", "company 1"],
        ["3", "rohith", "company 2"],
        ["4", "sridevi", "company 1"],
        ["5", "bobby", "company 1"]]
columns = ['ID', 'NAME', 'Company']
# creating a dataframe from the lists of data
df = spark.createDataFrame(data, columns)
df1=df.withColumn("salary",lit(30000)).show()
```

```python
# Add a column named salary from ID column with multiply of 2300

df.withColumn("salary", df.ID*2300).show()
```

## Add a column named details from name and company columns separated by '-'

```python
from pyspark.sql.functions import concat_ws

df.withColumn("Details", concat_ws("-", "NAME", "Company")).show()
```
